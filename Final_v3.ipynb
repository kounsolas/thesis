## Create functions that help me extract context from the sast report

````python
SNIPPET_DIR = "/content/vscode_demo"
SNIPPET_CONTEXT = 15

Sast_outputs = []
Numbered_code = []
vuln_snippets_info = []

def extract_snippet(filepath, line, context=SNIPPET_CONTEXT):
    with open(filepath, "r", encoding="utf-8", errors="ignore") as f:
        lines = f.readlines()
    start_line = max(0, int(line) - context - 1)
    end_line = min(len(lines), int(line) + context)
    return lines[start_line:end_line], start_line + 1

def number_whole_file(filepath):
    with open(filepath, "r", encoding="utf-8", errors="ignore") as f:
        lines = f.readlines()
    return "\n".join(f"{i+1:3} {l.rstrip()}" for i, l in enumerate(lines))

def parse_cwe_ids(cwe_ids):
    if not cwe_ids:
        return "CWE-UNKNOWN"
    return f"CWE-{str(cwe_ids[0]).strip()}"

def process_findings(vulnerabilities, severity_label):
    made, skipped = 0, 0

    for f in (vulnerabilities or []):
        file_abs = f.get("full_filename") or f.get("filename")
        if not file_abs or not isinstance(file_abs, str):
            skipped += 1
            continue

        # make relative path if under snippet dir
        file_rel = os.path.relpath(file_abs, SNIPPET_DIR) if file_abs.startswith(SNIPPET_DIR) else file_abs

        # ensure file exists
        if not os.path.isfile(file_abs):
            candidate = os.path.join(SNIPPET_DIR, file_rel)
            if os.path.isfile(candidate):
                file_abs = candidate
            else:
                print(f"[!] File not found: {file_abs}")
                skipped += 1
                continue

        line = f.get("line_number") or f.get("parent_line_number") or 1
        try:
            line = int(line)
        except Exception:
            line = 1

        rule = f.get("rule", {}) or {}
        cwe_id = parse_cwe_ids(rule.get("cwe_ids"))
        vuln_type = rule.get("title") or "Unknown"
        description = rule.get("description") or ""

        whole_numbered = number_whole_file(file_abs)



        sast_output = {
            "vulnerability_type": vuln_type,
            "cwe": cwe_id,
            "severity": severity_label.upper(),
            "description": description,
            "line": line,
            "file_path": file_rel
        }

        vuln_snippets_info.append({
            "file_path": file_rel,
            "file_abs": file_abs,
            "lang": vuln_code_ext,
            "sast_output": sast_output,
            "numbered_code_whole": whole_numbered,
        })
        made += 1

    return made, skipped

made_c, skip_c = process_findings(sast_critical, "CRITICAL")
made_h, skip_h = process_findings(sast_high, "HIGH")
made_m, skip_m = process_findings(sast_medium, "MEDIUM")
made_l, skip_l = process_findings(sast_low, "LOW")

# --- fallback: if NOTHING was found at all ---
total_findings = len(vuln_snippets_info)
if total_findings == 0:
    target_file = os.environ.get("TARGET_FILE")  # the one you scanned
    if target_file and os.path.isfile(target_file):
        whole_numbered = number_whole_file(target_file)
        file_rel = os.path.relpath(target_file, SNIPPET_DIR) if target_file.startswith(SNIPPET_DIR) else target_file

        vuln_snippets_info.append({
            "file_path": file_rel,
            "file_abs": target_file,
            "lang": vuln_code_ext,
            "sast_output": None,
            "numbered_code_whole": whole_numbered,
        })
    else:
        print("[!] No findings and TARGET_FILE is missing/unset.")

print("Total examples:", len(vuln_snippets_info))
````

## The model decides if the vulnerability is TP

````python
def build_tp_prompt() -> str:
    window_note = ""
    if isinstance(win_start, int) and isinstance(win_end, int):
        window_note = (
            f"\nThe snippet shows only lines {win_start}-{win_end} from the original file; "
            "line numbers are original."
        )

    has_sast = isinstance(sast_output, dict) and len(sast_output) > 0
    sast_flagged_line = vuln_line if isinstance(vuln_line, int) else None

    sast_type = None
    if has_sast:
        sast_type = (
            sast_output.get("vulnerability_type")
            or sast_output.get("title")
            or (sast_output.get("rule") or {}).get("title")
        )

    if has_sast:
        context_block = dedent(f"""\
        ## SAST report (JSON)
        {json.dumps(sast_output, indent=2)}
        SAST vulnerability type (if any): {sast_type}
        Flagged line (if any): {sast_flagged_line}
        {window_note}
        """)
        task_block = dedent("""\
        ## Task
        You must evaluate FOUR things:

        A) Existence: Does the code contain a real security vulnerability?
        B) Type correctness: Is the SAST vulnerability type correct for this code?
           - If wrong or incomplete, you must provide the correct type.
        C) CWE: Provide the best matching CWE in the form "CWE-<number>" (e.g., "CWE-89").
        D) Location: What are the correct vulnerable line numbers?
           - If you agree with the SAST flagged line (if provided), set line_match: true and include it in vuln_lines.
           - If you disagree or the flagged line is missing/invalid, set line_match: false and output corrected vuln_lines.

        Be strict and code-specific. Do not guess. If there isn’t enough evidence, set has_vulnerability: false.
        """)
        line_match_rule = (
            "Set line_match: true only if the SAST flagged line (if provided) matches the vulnerable operation "
            "(sink/source or inseparable part). Otherwise set line_match: false."
        )
        vuln_lines_rule = (
            "If has_vulnerability is false, vuln_lines MUST be []. "
            "If has_vulnerability is true and line_match is true and a flagged line exists, vuln_lines MUST include it. "
            "If has_vulnerability is true and line_match is false, vuln_lines MUST contain your corrected vulnerable lines."
        )
        type_rule = (
            "Set sast_type_correct: true only if the SAST vulnerability type matches what the code actually does. "
            "If false, set vulnerability_type to the correct type."
        )
        cwe_rule = (
            'If has_vulnerability is true, set cwe to the best matching CWE string like "CWE-89". '
            'If has_vulnerability is false, set cwe to "CWE-UNKNOWN".'
        )
    else:
        context_block = dedent(f"""\
        ## SAST report
        None (no findings were reported).
        {window_note}
        """)
        task_block = dedent("""\
        ## Task
        1) Decide if the code contains any real security vulnerability.
        2) If yes, identify the vulnerability type.
        3) If yes, provide the best matching CWE in the form "CWE-<number>" (e.g., "CWE-79").
        4) Output the exact vulnerable line numbers.
        5) Give a short, code-specific reason.

        Since there is no SAST report, you MUST set line_match: false and sast_type_correct: false.
        """)
        line_match_rule = "Because there is no SAST report, you MUST set line_match: false."
        vuln_lines_rule = (
            "If has_vulnerability is false, vuln_lines MUST be []. "
            "If has_vulnerability is true, vuln_lines MUST contain the vulnerable line numbers you identified."
        )
        type_rule = (
            "Because there is no SAST report, you MUST set sast_type_correct: false. "
            'If has_vulnerability is false, set vulnerability_type to "none".'
        )
        cwe_rule = (
            'Because there is no SAST report: if has_vulnerability is true, set cwe to the best matching "CWE-<number>"; '
            'if has_vulnerability is false, set cwe to "CWE-UNKNOWN".'
        )

    prompt = dedent(f"""\
    You are given a numbered code snippet. A SAST report may or may not be provided.
    Your job is to determine whether a real vulnerability exists, whether the SAST type is correct, the best CWE, and where it occurs.

    ## Code
    ```{lang}
    {numbered_code.strip()}
    ```

    {context_block}

    {task_block}

    ## Output Format (YAML)
    Return a YAML object matching this schema (no extra text):

    ```python
    class VulnGate(BaseModel):
        has_vulnerability: bool
        line_match: bool
        vuln_lines: List[int]         # original source line numbers from the snippet
        sast_type_correct: bool
        vulnerability_type: str       # if no vuln -> "none"
        cwe: str                      # e.g., "CWE-89"
        reason: str                   # ≤ 3 sentences, code-specific
    ```

    Rules:
    - If has_vulnerability is false:
      - vuln_lines MUST be []
      - vulnerability_type MUST be "none"
      - cwe MUST be "CWE-UNKNOWN"
    - {line_match_rule}
    - {vuln_lines_rule}
    - {type_rule}
    - {cwe_rule}
    - vuln_lines must refer to the line numbers shown in the snippet.

    Return only the YAML object.
    """)
    return prompt
````

# Create functions that help the patch generation process

## Entropy can be calculated as the negative log probability of each generated token

source: `Automated Program Repair in the Era of Large Pre-trained Language Models`

image.png


````python
def entropy_from_generate_outputs(output, sequences, input_len: int):
    """
    Υπολογίζει:
      - sum_nll: άθροισμα negative log-probs μόνο για τα generated tokens
      - avg_nll: μέσο negative log-prob ανά generated token
    """
    # sequences: [batch, prompt_len + gen_len]
    gen_ids = sequences[0, input_len:]           # οι παραγόμενοι tokens
    scores = output.scores                       # λίστα μήκους gen_len με logits
    logprobs = []

    for step, logits in enumerate(scores):
        # logits σχήμα [batch, vocab]; παίρνουμε το πρώτο batch
        lp = F.log_softmax(logits[0], dim=-1)
        tok_id = gen_ids[step]
        logprobs.append(lp[tok_id].item())

    if not logprobs:
        return float("inf"), float("inf")

    sum_nll = -sum(logprobs)
    avg_nll = sum_nll / len(logprobs)
    return sum_nll, avg_nll


def generate_multiple_yaml_patches(prompt, model, tokenizer, n, temperature, max_new_tokens,do_sample=True):
    patches = []
    entropies_sum = []
    entropies_avg = []

    system_message = (
        "You are a security coding assistant. Your goal is to generate safe, correct code patches in YAML format. "
        "Never output anything except valid YAML. The YAML must follow exactly the requested format. Be precise and concise."
    )

    for _ in range(n):
        # --- Build message
        messages = [
            {"role": "system", "content": system_message},
            {"role": "user", "content": prompt}
        ]

        # --- Tokenize
        text = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
        model_inputs = tokenizer([text], return_tensors="pt").to(model.device)

        if tokenizer.pad_token_id is None:
          tokenizer.pad_token_id = tokenizer.eos_token_id

        model.eval()
        with torch.inference_mode():
            output = model.generate(
                **model_inputs,
                do_sample=do_sample,
                temperature=temperature,
                top_p=0.95,
                max_new_tokens=max_new_tokens,
                return_dict_in_generate=True,
                output_scores=True,
                eos_token_id=tokenizer.eos_token_id,
                pad_token_id=tokenizer.pad_token_id,
                use_cache=True,
            )

        input_len = model_inputs.input_ids.shape[-1]
        sequences = output.sequences
        output_ids = sequences[0, input_len:]
        generated_text = tokenizer.decode(output_ids, skip_special_tokens=True).strip()

        # entropy μόνο στα generated tokens
        sum_nll, avg_nll = entropy_from_generate_outputs(output, sequences, input_len)

        patches.append(generated_text)
        entropies_sum.append(float(sum_nll))
        entropies_avg.append(float(avg_nll))

    return patches, entropies_sum, entropies_avg
````

````python
def extract_import_block(file_path, max_lines=80):
    lines = code

    i = 0
    while i < len(lines) and (lines[i].lstrip().startswith("//") or lines[i].lstrip().startswith("/*") or lines[i].strip()=="" or lines[i].strip().startswith("*")):
        i += 1

    import_re = re.compile(r'^\s*(import\s|export\s+.*\sfrom\s|const\s+\w+\s*=\s*require\s*\()')
    block = []
    while i < len(lines) and len(block) < max_lines:
        if import_re.match(lines[i]):
            block.append(lines[i].rstrip("\n"))
            i += 1
        else:
            break

    return "\n".join(block)
````

# Functions that validate the patches and selects the best ones

````python
def validate_patch_obj(obj):
    """
    Validates that the patch has the correct format
    """
    # Check that the response contains a patch field
    if not isinstance(obj, dict) or "patch" not in obj:
        return False, "root must contain 'patch'"
    p = obj["patch"]

    # Check that patch is a dictionary
    if not isinstance(p, dict):
        return False, "'patch' must be a mapping"

    # Check that the dictionary `patch` has the fields `code` and `replace_lines`
    if "replace_lines" not in p or "code" not in p:
        return False, "missing 'replace_lines' or 'code'"
    rl = p["replace_lines"]

    #Check that replace_lines is [int,int]
    if not (isinstance(rl, list) and len(rl) == 2 and all(isinstance(x, int) for x in rl)):
        return False, "'replace_lines' must be [int, int]"
    if rl[0] > rl[1]:
        return False, "replace_lines start>end"
    if not isinstance(p["code"], str) or not p["code"].strip():
        return False, "'code' must be a non-empty string"
    return True, "ok"
````

````python
def select_topN(patches: List[Dict[str, Any]], n: int) -> List[Dict[str, Any]]:


    def covers_vuln(p):
        """
        Check if the patch covers the vulnerable line
        """
        a, b = p["replace_lines"]
        for vuln_l in vuln_line:
          if a <= vuln_l <= b:
              return True
        else:
            print(f"The patch does NOT cover the vulnerable line. Skip ...\n")
            return False


    def inside_window(p):
        """
        Check if the patch is inside the context window
        """
        if win_start is None or win_end is None:
            return True
        a, b = p["replace_lines"]
        if (a >= win_start) and (b <= win_end):
            return True
        else:
            print(f"Thepatch is outside the context window. Skip ...")
            return False

    # 1) φίλτρο: καλύπτουν τη vuln_line
    cand = [p for p in patches if covers_vuln(p)]
    # 2) φίλτρο: μέσα στο window
    cand = [p for p in cand if inside_window(p)]

    if not cand:
        print(f"All the filters failed ...")
        cand = patches[:]

    # 3) ταξινόμηση:  χαμηλότερο sum entropy, avg entropy
    cand.sort(key=lambda p: (p["sum_entropy"], p["avg_entropy"]))
    # print(f"---PATCH WITH MOST ENTROPY-------------\n")
    # pprint(cand[-1])
    return cand[:n]
````

````python
def extract_yaml(s: str) -> str:
    s = s.strip()
    if s.startswith("```yaml"):
        s = s[len("```yaml"):].strip()
        if s.endswith("```"):
            s = s[:-3].strip()
    if s.startswith("---"):
        return s  # ok είτε με --- είτε χωρίς
    return s
````

````python
_BAD_BLOCK_RE = re.compile(r'(?m)^\s*[^:#\n]+:\s*\|\s*(?:0|\d{2,})\s*$')
# επιτρέπουμε ΜΟΝΟ |, |-, |+, |>... με το πολύ 1 ψηφίο (1-9). Το 0 ή 2+ ψηφία -> invalid.

def load_yaml_patch_strict(yaml_text: str):
    if _BAD_BLOCK_RE.search(yaml_text):
        return None, "invalid YAML block scalar indicator like '|36'"
    try:
        obj = yaml.safe_load(yaml_text)
    except yaml.YAMLError as e:
        return None, f"yaml parse error: {e}"
    return obj, "ok"
````

# Generation Loop


````python
def get_yaml_patches(
    prompt: str,
    temperature: float,
    max_new_tokens: int,
    n_valid: int = 3,
    per_round: int = 4,
    max_rounds: int = 5,
    remove_duplicates: bool = True,
    do_sample = True
) -> List[Dict[str, Any]]:
    """
    Γεννά και επιστρέφει n_valid έγκυρα patches (validate_patch_obj).
    Αν κάποιο patch δεν περνά validation, ξαναπροσπαθεί με νέες γεννήτριες
    μέχρι να βρει αρκετά ή να εξαντλήσει γύρους.
    Στο τέλος επιλέγει τα καλύτερα n_valid με select_topN.
    """
    valid_patches: List[Dict[str, Any]] = []
    seen = set()

    for round in range(max_rounds):
        print(f"=======Generation Round : {round + 1}=======\n")
        gen_result = generate_multiple_yaml_patches(
            model=model,
            tokenizer=tokenizer,
            prompt=prompt,
            n=per_round,
            temperature=temperature,
            max_new_tokens=max_new_tokens,
            do_sample=do_sample
        )

        try:
          yaml_texts, entropies_sum, entropies_avg = gen_result
        except ValueError as e:
          raise RuntimeError(
              "generate_multiple_yaml_patches must return exactly (patches, entropies_sum, entropies_avg)"
          ) from e


        # sort των indices με βάση avg_nll ή sum_nll
        idxs = list(range(len(yaml_texts)))

        for i in idxs:
            print(f"{i+1}.")
            raw = yaml_texts[i]
            try:
                cleaned = extract_yaml(raw)  # καθαρίζει πιθανά ```yaml fences
                obj, msg = load_yaml_patch_strict(cleaned)
                if obj is None:
                  print(f"There was a syntax YAML error. {msg}. Skip this patch ...")
                  continue
                ok, msg = validate_patch_obj(obj)
                if not ok:
                    print(f"Validation failed: {msg}")
                    continue

                p = obj["patch"]

                # ADD KEY FIELDS IN THE PATCH DICTIONARY
                p["sum_entropy"] = float(entropies_sum[i])
                p["avg_entropy"] = float(entropies_avg[i])

                if remove_duplicates:
                    sig = (tuple(p["replace_lines"]), p["code"].strip())
                    if sig in seen:
                        print("Duplicate patch. Continue ...")
                        continue
                    seen.add(sig)

                valid_patches.append(p)


            except Exception:
                continue

    if not valid_patches:
        return []
    print(f"Selecting top {n_valid} patches from total {len(valid_patches)} patches ...")
    # κράτα τα καλύτερα n_valid με βάση vuln_line/window/entropy/span
    topN = select_topN(valid_patches, n=n_valid)
    return topN
````

````python
t0 = time.perf_counter()

yaml_patches = get_yaml_patches(
    prompt=yaml_prompt,
    n_valid=3,
    temperature=0.9,  #0.9 - 0.95 best
    max_new_tokens=1024,
    per_round=4,
    max_rounds=3,
    remove_duplicates=True,
    do_sample=True
)

elapsed = time.perf_counter() - t0
m, s = divmod(elapsed, 60)
print(f"get_yaml_patches took {int(m)}m {s:.2f}s")
pprint(yaml_patches)
````

# The model selects the best patch

````python
def build_best_patch_prompt() -> str:

  """
    Δίνει στο μοντέλο όλα όσα χρειάζεται για να επιλέξει *ένα* από τα δοθέντα patches,
  """
  patches_yaml = yaml.dump(yaml_valid_patches, sort_keys=False)


  prompt_template = """You are a security reviewer. Choose the best fix **only among the given patches**.

  ## Context
  - Vulnerable line in original file: {vuln_line}
  - Snippet window: [{win_start}, {win_end}]

  ## Imports (for context)
  ```{lang}
  {imports}
  ```

  ## Vulnerable code:
  ```{lang}
  {numbered_window}
  ```

  ## SAST report:
  {sast_output}

  ## Self-reflection on the problem:
  {self_reflection}

  ## Candidate patches (YAML list):
  {yaml_valid_patches}

  ## Optional tests (guidance)
  {tests_yaml}

  ## Security Policy - {policy_name}
  Allowed:
  {allowed_list}
  Disallowed:
  {disallowed_list}

  ## Task
  - Disqualify any patch that violates any **Disallowed** item.
  - Prefer patches that follow the **Allowed** items.
  - Fully eliminate the vulnerability .
  - Preserve original behavior (no extra logic/side-effects)..
  - Be syntactically valid with the given imports.

  ## Output Format (YAML)
  Return ONE YAML object, no code fences, starting with ---. Fields:
  - `chosen_index`: the 0-based index of the chosen candidate from the list above
  - `why`: short rationale (security + correctness)
  - `flow`: 2–4 bullets with reasoning
  Start your YAML with --- and do not include any extra text.

  ## Example shape (do not copy literally)
  chosen_index: 0
  why: |
  [short rationale]
  flow:
  - |
    [reasoning bullet]
  - |
    [reasoning bullet]

  """
  return prompt_template.format(sast_output = sast_output,
                                numbered_window = numbered_window,
                                self_reflection=self_reflection,
                                yaml_valid_patches=patches_yaml.strip(),
                                lang=lang,
                                vuln_line=vuln_line,
                                imports = imports,
                                win_start = win_start,
                                win_end = win_end,
                                tests_yaml = tests_yaml,
                                policy_name = policy["name"],
                                allowed_list = '\n'.join([f"- {a}" for a in policy.get("allowed", [])]),
                                disallowed_list = '\n'.join([f"- {d}" for d in policy.get("disallowed", [])]),
                                )

````

````python
best_patch_prompt = build_best_patch_prompt()
print(best_patch_prompt)
````

````python
messages = [
    {"role": "system", "content": "You are a security reviewer tasked with selecting the single best code patch from a provided list of candidates. Your selection must be based on the vulnerability details, security policy, and test guidance. Prioritize patches that are secure, correct and adhere to policy. Your output MUST be a YAML object containing the index of the chosen patch and your reasoning. Return ONLY the YAML object, no extra text or markdown."},
    {"role": "user", "content":best_patch_prompt}
]

text = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
model_inputs = tokenizer([text], return_tensors="pt").to(model.device)

# --- Generate
with torch.no_grad():
    output = model.generate(
        **model_inputs,
        do_sample=True,   #do_sample = False παιρνεις το πιο πιθανο token (greedy approac)
        temperature=0.9,
        top_p=0.95,
        max_new_tokens=512,
        return_dict_in_generate=True,
        output_scores=True
    )

# --- Extract output ids and text
output_ids = output.sequences[0][model_inputs.input_ids.shape[-1]:]
generated_text = tokenizer.decode(output_ids, skip_special_tokens=True).strip()
````

````python
obj = yaml.safe_load(extract_yaml(generated_text))
pprint(obj)
````
